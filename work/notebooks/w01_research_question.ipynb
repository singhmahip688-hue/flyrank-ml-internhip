Search question: Which pages should a content reviewer look at first this week, because they're stale, losing search visibility, or under-performing for how much traffic potential they have?

Decision this improves: where a content team spends its limited review/editing time each week.

Action taken: a reviewer opens the top-ranked pages first and decides whether to refresh, expand, prune, or leave each one.
